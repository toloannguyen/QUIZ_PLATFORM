```python
def to_lowercase(text):
    if pd.isnull(text):
        return ""
    return str(text).lower()
```

```python
import re

def normalize_spaces(text):
    return re.sub(r'\s+', ' ', text)
```

```python
print(normalize_spaces("nước    sôi ở   100 độ"))
# → "nước sôi ở 100 độ"
```

```python
def clean_whitespace(text):
    if pd.isnull(text):
        return ""
    text = str(text).strip()          # xóa đầu/cuối trước
    text = re.sub(r'\s+', ' ', text)  # chuẩn hóa khoảng trắng giữa
    return text
```

```python
print(clean_whitespace("   nước   sôi ở  100 độ C   "))
# → "nước sôi ở 100 độ C"
```

Bước 3.2: Viết hàm xóa dấu câu cơ bản
Giải thích:
string.punctuation — 1 chuỗi có sẵn trong thư viện string, liệt kê hết các ký tự dấu câu tiếng Anh chuẩn (bao gồm cả %, - — nên nếu dùng nguyên bản này sẽ xóa luôn cả 2 ký tự bạn có thể muốn giữ theo bảng ở Bước 3.1).
str.maketrans('', '', string.punctuation) — tạo ra 1 "bảng chuyển đổi" (translation table): 2 tham số đầu rỗng nghĩa là không thay ký tự nào bằng ký tự khác, tham số thứ 3 là danh sách ký tự cần xóa hẳn.
.translate(translator) — áp bảng chuyển đổi đó lên chuỗi.

```python
import string

def remove_punctuation(text):
    # string.punctuation chứa: !"#$%&'()*+,-./:;<=>?@[\]^_`{|}~
    translator = str.maketrans('', '', string.punctuation)
    return text.translate(translator)
```

```python
import string

def remove_punctuation(text):
    if pd.isnull(text):
        return ""
    # Loại bỏ dấu câu nhưng GIỮ LẠI ký tự %
    punctuation_to_remove = string.punctuation.replace('%', '')
    translator = str.maketrans('', '', punctuation_to_remove)
    return str(text).translate(translator)
```

```python
test_cases = [
    "Nước sôi ở 100 độ C, đúng không?",
    "Tỷ lệ tăng 20% - rất đáng kể!",
    "Câu hỏi: Trái Đất có hình gì???",
]

for t in test_cases:
    print(f"Gốc:  {t}")
    print(f"Sạch: {remove_punctuation(t)}")
    print()
```

```python
def clean_text(text):
    if pd.isnull(text):
        return ""
    
    text = str(text)
    text = to_lowercase(text)
    text = remove_punctuation(text)
    text = clean_whitespace(text)
    
    return text
```

```python
test_cases = [
    "   Nước SÔI ở 100 Độ C,   đúng không???   ",
    "TỶ  Lệ   tăng 20%!!! - Rất Đáng   Kể",
    "",
    None,
]

for t in test_cases:
    print(f"Gốc:  {repr(t)}")
    print(f"Sạch: {repr(clean_text(t))}")
    print()
```
