Bước 3.2: Viết hàm xóa dấu câu cơ bản
Giải thích:
string.punctuation — 1 chuỗi có sẵn trong thư viện string, liệt kê hết các ký tự dấu câu tiếng Anh chuẩn (bao gồm cả %, - — nên nếu dùng nguyên bản này sẽ xóa luôn cả 2 ký tự bạn có thể muốn giữ theo bảng ở Bước 3.1).
str.maketrans('', '', string.punctuation) — tạo ra 1 "bảng chuyển đổi" (translation table): 2 tham số đầu rỗng nghĩa là không thay ký tự nào bằng ký tự khác, tham số thứ 3 là danh sách ký tự cần xóa hẳn.
.translate(translator) — áp bảng chuyển đổi đó lên chuỗi.

```python
import string

def remove_punctuation(text):
    # string.punctuation chứa: !"#$%&'()*+,-./:;<=>?@[\]^_`{|}~
    translator = str.maketrans('', '', string.punctuation)
    return text.translate(translator)
```

```python
import string

def remove_punctuation(text):
    if pd.isnull(text):
        return ""
    # Loại bỏ dấu câu nhưng GIỮ LẠI ký tự %
    punctuation_to_remove = string.punctuation.replace('%', '')
    translator = str.maketrans('', '', punctuation_to_remove)
    return str(text).translate(translator)
```

```python
test_cases = [
    "Nước sôi ở 100 độ C, đúng không?",
    "Tỷ lệ tăng 20% - rất đáng kể!",
    "Câu hỏi: Trái Đất có hình gì???",
]

for t in test_cases:
    print(f"Gốc:  {t}")
    print(f"Sạch: {remove_punctuation(t)}")
    print()
```

```python
def clean_text(text):
    if pd.isnull(text):
        return ""
    
    text = str(text)
    text = to_lowercase(text)
    text = remove_punctuation(text)
    text = clean_whitespace(text)
    
    return text
```

```python
# Gộp TOÀN BỘ câu (cả reference và student answer) để fit 1 lần duy nhất
all_texts = [
    "nước sôi ở 100 độ c",
    "nước sôi khi đun nóng đủ lâu",
    "trái đất quay quanh mặt trời",
    "mặt trời quay quanh trái đất"
]

vectorizer3 = TfidfVectorizer()
vectorizer3.fit(all_texts)   # học từ vựng 1 LẦN từ TOÀN BỘ dữ liệu

# Sau đó, transform từng câu riêng lẻ khi cần so sánh
ref_vector = vectorizer3.transform(["nước sôi ở 100 độ c"])
student_vector = vectorizer3.transform(["nước sôi khi đun nóng đủ lâu"])

print(ref_vector.shape, student_vector.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity_score = cosine_similarity(ref_vector, student_vector)
print(similarity_score)
```

```python
def compute_similarity(text1, text2, vectorizer):
    vec1 = vectorizer.transform([text1])
    vec2 = vectorizer.transform([text2])
    score = cosine_similarity(vec1, vec2)
    return score[0][0]
```

```python
test_corpus = [
    "nước sôi ở 100 độ c",
    "nước sôi khi đun nóng đủ lâu",
    "nước không sôi ở 100 độ c",
    "con mèo là động vật có vú",
    "h2o bắt đầu sôi tại 100 độ c"
]

test_vectorizer = TfidfVectorizer()
test_vectorizer.fit(test_corpus)
```

```python
score1 = compute_similarity(
    "nước sôi ở 100 độ c",
    "nước sôi ở 100 độ c",
    test_vectorizer
)
print("Giống hệt nhau:", score1)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def compute_semantic_similarity(text1, text2):
    vec1 = encode_text(text1).reshape(1, -1)
    vec2 = encode_text(text2).reshape(1, -1)
    score = cosine_similarity(vec1, vec2)
    return score[0][0]
```

```python
THRESHOLD_VERY_GOOD = 0.70   # >= 0.70 → Very Good
THRESHOLD_PARTIAL = 0.45     # >= 0.45 và < 0.70 → Partially Relevant
                              # < 0.45 → Not Related
```

```python
def classify_by_threshold(similarity_score, threshold_very_good=0.70, threshold_partial=0.45):
    if similarity_score >= threshold_very_good:
        return "Very Good"
    elif similarity_score >= threshold_partial:
        return "Partially Relevant"
    else:
        return "Not Related"
```

```python
def has_asymmetric_negation(text1, text2):
    negation_words = ["không", "chẳng", "chả"]
    # Các từ ghép chứa "không"/"chẳng" nhưng KHÔNG mang nghĩa phủ định
    exception_words = ["không khí", "không gian", "không dây", "chẳng hạn"]
    
    def count_negation(text):
        text_copy = text
        for exc in exception_words:
            text_copy = text_copy.replace(exc, "")
        words = text_copy.split()
        return any(word in words for word in negation_words)
    
    text1_has_negation = count_negation(text1)
    text2_has_negation = count_negation(text2)
    
    return text1_has_negation != text2_has_negation
```

```python
antonym_pairs = [
    ("tăng", "giảm"),
    ("có", "không có"),
    ("đúng", "sai"),
    ("nhanh", "chậm"),
    ("nóng", "lạnh"),
    ("cao", "thấp"),
    ("lớn", "nhỏ"),
    ("nhiều", "ít"),
    ("trước", "sau"),
    ("trong", "ngoài"),
]
```

```python
def has_antonym_conflict(text1, text2):
    words1 = set(text1.split())
    words2 = set(text2.split())
    
    for word_a, word_b in antonym_pairs:
        # Trường hợp: câu 1 có từ A, câu 2 có từ B (hoặc ngược lại) — dấu hiệu đối nghĩa
        if (word_a in words1 and word_b in words2) or (word_b in words1 and word_a in words2):
            return True
    
    return False
```

```python
print(has_antonym_conflict(
    "áp suất khí quyển giảm dần khi lên cao",
    "áp suất khí quyển tăng dần khi lên cao"
))  # kỳ vọng: True
```

```python
def has_conflict_signal(text1, text2):
    return has_asymmetric_negation(text1, text2) or has_antonym_conflict(text1, text2)
```

```python
def classify_with_conflict_check(ref_text, student_text, similarity_score,
                                   threshold_very_good=0.72, threshold_partial=0.40):
    # Bước 1: phân loại cơ bản theo ngưỡng (như Task 2)
    base_label = classify_by_threshold(similarity_score, threshold_very_good, threshold_partial)
    
    # Bước 2: kiểm tra tín hiệu xung đột (phủ định/đối nghĩa)
    conflict_detected = has_conflict_signal(ref_text, student_text)
    
    # Bước 3: nếu có xung đột VÀ nhãn cơ bản không phải "Not Related" → hạ xuống "Not Related"
    if conflict_detected and base_label != "Not Related":
        final_label = "Not Related"
    else:
        final_label = base_label
    
    return final_label, conflict_detected
```

```python
def has_antonym_conflict(text1, text2):
    words1 = set(text1.split())
    words2 = set(text2.split())
    
    for word_a, word_b in antonym_pairs:
        # Chỉ báo xung đột nếu từ A CHỈ xuất hiện ở 1 câu, từ B CHỈ xuất hiện ở câu kia
        # (loại trừ trường hợp cả 2 từ cùng xuất hiện chung 1 câu — dấu hiệu mô tả quan hệ, không phải mâu thuẫn)
        case1 = (word_a in words1 and word_b in words2) and (word_b not in words1) and (word_a not in words2)
        case2 = (word_b in words1 and word_a in words2) and (word_a not in words1) and (word_b not in words2)
        if case1 or case2:
            return True
    
    return False
```

```python
def classify_answer(ref_text, student_text, threshold_very_good=0.72, threshold_partial=0.40):
    # Bước 1: kiểm tra câu trả lời trống trước tiên
    if is_empty_answer(student_text):
        return {
            "label": "Not Related",
            "similarity_score": 0.0,
            "conflict_detected": False,
            "reason": "Câu trả lời trống"
        }
    
    # Bước 2: tính semantic similarity
    similarity_score = compute_semantic_similarity(ref_text, student_text)
    
    # Bước 3: phân loại theo ngưỡng
    base_label = classify_by_threshold(similarity_score, threshold_very_good, threshold_partial)
    
    # Bước 4: kiểm tra xung đột (phủ định/đối nghĩa)
    conflict_detected = has_conflict_signal(ref_text, student_text)
    
    # Bước 5: hạ nhãn nếu phát hiện xung đột
    if conflict_detected and base_label != "Not Related":
        final_label = "Not Related"
        reason = "Phát hiện phủ định/đối nghĩa bất đối xứng, hạ nhãn để an toàn"
    else:
        final_label = base_label
        reason = None
    
    return {
        "label": final_label,
        "similarity_score": round(float(similarity_score), 4),
        "conflict_detected": conflict_detected,
        "reason": reason
    }
```

```python
def chunk_text(text, chunk_size=500, chunk_overlap=50):
    chunks = []
    start = 0
    text_length = len(text)
    
    while start < text_length:
        end = start + chunk_size
        chunk = text[start:end]
        chunks.append(chunk)
        
        start += chunk_size - chunk_overlap
    
    return chunks
```

```python
def create_chunks_with_metadata(pages_data, source_filename, chunk_size=500, chunk_overlap=50):
    all_chunks = []
    chunk_index = 0
    
    for page in pages_data:
        page_chunks = chunk_text(page['clean_text'], chunk_size, chunk_overlap)
        
        for c in page_chunks:
            all_chunks.append({
                "chunk_id": f"{source_filename}_chunk_{chunk_index}",
                "source_file": source_filename,
                "page_number": page['page_number'],
                "chunk_index": chunk_index,
                "text": c,
                "char_count": len(c)
            })
            chunk_index += 1
    
    return all_chunks
```

```python
def add_document_to_db(chunks, collection, model):
    ids = [c['chunk_id'] for c in chunks]
    documents = [c['text'] for c in chunks]
    metadatas = [
        {
            "source_file": c['source_file'],
            "page_number": c['page_number'],
            "chunk_index": c['chunk_index']
        }
        for c in chunks
    ]

    embeddings = model.encode(documents).tolist()

    collection.upsert(
        ids=ids,
        documents=documents,
        embeddings=embeddings,
        metadatas=metadatas
    )

    return {
        "status": "success",
        "chunks_added": len(chunks),
        "total_in_collection": collection.count()
    }
```

```python
def retrieve_relevant_chunks(student_answer, collection, model, top_k=3):
    query_embedding = model.encode([student_answer]).tolist()
    
    results = collection.query(
        query_embeddings=query_embedding,
        n_results=top_k
    )
    
    chunks = []
    for i in range(len(results['ids'][0])):
        distance = results['distances'][0][i]
        similarity = 1 - distance
        chunks.append({
            "chunk_id": results['ids'][0][i],
            "text": results['documents'][0][i],
            "similarity": round(similarity, 4),
            "page_number": results['metadatas'][0][i]['page_number'],
            "source_file": results['metadatas'][0][i]['source_file']
        })
    
    return chunks
```

```python
def evaluate_essay_answer(student_answer, collection, model, top_k=3, 
                            threshold_very_good=0.72, threshold_partial=0.40,
                            min_relevance_threshold=0.30):
    # Bước 1: kiểm tra câu trả lời trống trước tiên
    if is_empty_answer(student_answer):
        return {
            "label": "Not Related",
            "similarity_score": 0.0,
            "conflict_detected": False,
            "reason": "Câu trả lời trống",
            "reference_chunks": []
        }
    
    # Bước 2: truy vấn top chunks liên quan
    chunks = retrieve_relevant_chunks(student_answer, collection, model, top_k)
    
    # Bước 3: kiểm tra có tìm được chunk nào đủ liên quan không
    if not chunks or chunks[0]['similarity'] < min_relevance_threshold:
        return {
            "label": "Not Related",
            "similarity_score": chunks[0]['similarity'] if chunks else 0.0,
            "conflict_detected": False,
            "reason": "Không tìm thấy nội dung đủ liên quan trong tài liệu tham khảo",
            "reference_chunks": chunks
        }
    
    # Bước 4: dùng chunk liên quan nhất để chấm điểm qua classify_answer()
    best_chunk = chunks[0]
    classification = classify_answer(
        best_chunk['text'], 
        student_answer, 
        threshold_very_good, 
        threshold_partial
    )
    
    # Bước 5: đính kèm thông tin nguồn tham khảo vào kết quả cuối
    classification['reference_chunks'] = chunks
    classification['primary_source'] = {
        "page_number": best_chunk['page_number'],
        "source_file": best_chunk['source_file']
    }
    
    return classification
```

```python
def classify_answer(ref_text, student_text, threshold_very_good=0.72, threshold_partial=0.40):
    if is_empty_answer(student_text):
        return {
            "label": "Not Related",
            "similarity_score": 0.0,
            "conflict_detected": False,
            "reason": "Câu trả lời trống"
        }
    
    similarity_score = compute_semantic_similarity(ref_text, student_text)
    base_label = classify_by_threshold(similarity_score, threshold_very_good, threshold_partial)
    conflict_detected = has_conflict_signal(ref_text, student_text)
    
    if conflict_detected and base_label != "Not Related":
        final_label = "Not Related"
        reason = "Phát hiện phủ định/đối nghĩa bất đối xứng, hạ nhãn để an toàn"
    else:
        final_label = base_label
        reason = None
    
    return {
        "label": final_label,
        "similarity_score": round(float(similarity_score), 4),
        "conflict_detected": conflict_detected,
        "reason": reason
    }
```

```python
def evaluate_essay_answer(student_answer, collection, model, top_k=3, 
                            threshold_very_good=0.72, threshold_partial=0.40,
                            min_relevance_threshold=0.30):
    if is_empty_answer(student_answer):
        return {
            "label": "Not Related",
            "similarity_score": 0.0,
            "conflict_detected": False,
            "reason": "Câu trả lời trống",
            "reference_chunks": []
        }
    
    chunks = retrieve_relevant_chunks(student_answer, collection, model, top_k)
    
    if not chunks or chunks[0]['similarity'] < min_relevance_threshold:
        return {
            "label": "Not Related",
            "similarity_score": chunks[0]['similarity'] if chunks else 0.0,
            "conflict_detected": False,
            "reason": "Không tìm thấy nội dung đủ liên quan trong tài liệu tham khảo",
            "reference_chunks": chunks
        }
    
    best_chunk = chunks[0]
    
    # Không dùng classify_answer() (có kiểm tra xung đột) — chỉ dùng ngưỡng thuần túy
    similarity_score = best_chunk['similarity']
    label = classify_by_threshold(similarity_score, threshold_very_good, threshold_partial)
    
    return {
        "label": label,
        "similarity_score": round(float(similarity_score), 4),
        "conflict_detected": None,
        "reason": "Lớp kiểm tra phủ định không áp dụng cho tài liệu dài (độ tin cậy thấp)",
        "reference_chunks": chunks,
        "primary_source": {
            "page_number": best_chunk['page_number'],
            "source_file": best_chunk['source_file']
        }
    }
```
